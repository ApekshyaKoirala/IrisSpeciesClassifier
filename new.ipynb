```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
df=pd.read_csv("iris.csv")
df.info()
```

```python
print(df.isnull().sum())
```

```python
names= df['Species'].value_counts().index
counts= df['Species'].value_counts().values
```

```python
plt.Figure(figsize=(6,4))
plt.bar(x=names,height=counts)
plt.title('Iris classification')
plt.xlabel('Species')
plt.ylabel('Count')
plt.savefig('plot.png')
plt.show()
```

```python
X=df.drop(columns=["Species"])
y=df['Species']
```

```python
print(df['Species'].value_counts())
```

```python
print(X.shape)
```

```python
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=123)
```

```python
X_train.head()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score,classification_report
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

```python
knn=KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train,y_train)
y_pred=knn.predict(X_test)
print('accuracy=', accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
cm=confusion_matrix(y_test,y_pred)
cm_plot=ConfusionMatrixDisplay(cm,display_labels=df['Species'].unique())

cm_plot.plot()
plt.savefig("confusionmatrix_png")
```

```python
test_pred=knn.predict([[5.1,3.5,1.4,0.2]])
print(test_pred[0])
```
